## What is a vector?

A vector is an ordered list of numbers. In linear algebra, vectors represent points or directions in space.

- A **2D vector** like `[3, 4]` represents a point (or arrow) in 2D space
- A **3D vector** like `[1, 2, 3]` represents a point in 3D space
- An **n-dimensional vector** is just a list of n numbers

In Python, we use **NumPy arrays** to represent vectors.

```python
import numpy as np
import matplotlib.pyplot as plt

print('NumPy version:', np.__version__)
```

## 1. Creating Vectors

```python
# Create vectors
v = np.array([3, 4])        # 2D vector
w = np.array([1, 2, 3])     # 3D vector

print('v =', v)
print('w =', w)
print('Shape of v:', v.shape)
print('Shape of w:', w.shape)
```

## 2. Vector Addition

Add vectors element-by-element. Both vectors must have the same dimensions.

$$\mathbf{a} + \mathbf{b} = [a_1 + b_1,\ a_2 + b_2]$$

```python
a = np.array([1, 2])
b = np.array([3, 1])

result = a + b
print('a =', a)
print('b =', b)
print('a + b =', result)
```

```python
# Visualize vector addition
fig, ax = plt.subplots(figsize=(6, 6))

origin = np.array([0, 0])

ax.quiver(*origin, *a, angles='xy', scale_units='xy', scale=1, color='blue', label='a')
ax.quiver(*a,      *b, angles='xy', scale_units='xy', scale=1, color='green', label='b (from tip of a)')
ax.quiver(*origin, *result, angles='xy', scale_units='xy', scale=1, color='red', label='a + b')

ax.set_xlim(-1, 6)
ax.set_ylim(-1, 5)
ax.set_aspect('equal')
ax.grid(True)
ax.legend()
ax.set_title('Vector Addition')
plt.tight_layout()
plt.show()
```

## 3. Scalar Multiplication

Multiplying a vector by a scalar stretches (or shrinks) it.

$$c \cdot \mathbf{v} = [c \cdot v_1,\ c \cdot v_2]$$

```python
v = np.array([2, 1])

print('v       =', v)
print('2 * v   =', 2 * v)    # stretch
print('0.5 * v =', 0.5 * v)  # shrink
print('-1 * v  =', -1 * v)   # flip direction
```

## 4. Vector Magnitude (Length)

The magnitude (or norm) of a vector is its length:

$$\|\mathbf{v}\| = \sqrt{v_1^2 + v_2^2 + \cdots + v_n^2}$$

For `v = [3, 4]`, we expect `||v|| = 5` (the classic 3-4-5 right triangle).

```python
v = np.array([3, 4])

# Method 1: manual calculation
magnitude_manual = np.sqrt(np.sum(v**2))

# Method 2: NumPy built-in
magnitude_numpy = np.linalg.norm(v)

print('v =', v)
print('||v|| (manual) =', magnitude_manual)
print('||v|| (numpy)  =', magnitude_numpy)
```

## 5. Unit Vectors

A unit vector has magnitude exactly 1. To normalize any vector:

$$\hat{v} = \frac{\mathbf{v}}{\|\mathbf{v}\|}$$

```python
v = np.array([3, 4])

v_hat = v / np.linalg.norm(v)

print('v     =', v,      '  magnitude =', np.linalg.norm(v))
print('v_hat =', v_hat,  '  magnitude =', np.linalg.norm(v_hat))
```

## 6. Dot Product

The dot product of two vectors produces a **scalar**:

$$\mathbf{a} \cdot \mathbf{b} = a_1 b_1 + a_2 b_2 + \cdots = \|\mathbf{a}\| \|\mathbf{b}\| \cos\theta$$

Key insight: **the dot product tells you how much two vectors point in the same direction.**
- Positive → same general direction
- Zero → perpendicular (orthogonal)
- Negative → opposite directions

```python
a = np.array([1, 0])   # pointing right
b = np.array([0, 1])   # pointing up
c = np.array([1, 1])   # pointing up-right

print('a · b (perpendicular):', np.dot(a, b))  # should be 0
print('a · c (same-ish dir):', np.dot(a, c))   # should be positive
print('a · -a (opposite):', np.dot(a, -a))     # should be negative
```

```python
# Compute the angle between two vectors using the dot product
def angle_between(v1, v2, degrees=True):
    cos_theta = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
    cos_theta = np.clip(cos_theta, -1, 1)  # guard against floating point errors
    angle_rad = np.arccos(cos_theta)
    return np.degrees(angle_rad) if degrees else angle_rad

a = np.array([1, 0])
b = np.array([1, 1])

print(f'Angle between {a} and {b}: {angle_between(a, b):.1f} degrees')
```
